```python

# Data
Q_in = 1  # inflow rate
A_t = 1 # cross section area of the tank (m^2)
g = 9.8 # gravity
A_out = 0.1  # cross section area of the outflow rate
h_0 = 0   # initial height of the water level
T = 100   # Time steps
N = 1000    # Total number of N cycle
dt = T/N    # step time period

```

```python
# Outflow rate

def Q_out(h):
  return A_out* (2*g*h)**0.5
```

```python
  # List to store the solutions

  times= (N+1)*[0]
  h=(N+1)*[0]
  h[0]= h_0
```

```python
#loop
for n in range(N):
    dh = dt*(Q_in-Q_out(h[n]))/A_t
    h[n+1]= h[n] + dh
    times[n+1]= (n+1) *dt
```

```python
#plot
import matplotlib.pyplot as plt
plt.plot(times, h)
plt.grid
plt.xlabel('t')
plt.ylabel('h')
```
